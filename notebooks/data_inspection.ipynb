```python
missing = df.isnull().sum()
missing_percent = (missing / len(df)) * 100

for col in df.columns:
    if missing[col] > 0:
        print(f"{col}: {missing[col]} ({missing_percent[col]:.1f}%)")
```

```python
if 'lyrics' in df.columns:
    total = len(df)
    null_lyrics = df['lyrics'].isnull().sum()
    empty_lyrics = (df['lyrics'] == '').sum()
    false_lyrics = (df['lyrics'] == 'False').sum()

    invalid_total = null_lyrics + empty_lyrics + false_lyrics
    valid_lyrics = total - invalid_total

    print("\nLYRICS DETAIL:")
    print(f"VALID: {valid_lyrics} ({valid_lyrics/total*100:.1f}%)")
    print(f"INVALID: {invalid_total} ({invalid_total/total*100:.1f}%)")
```
